```python
import cv2
vidcap = cv2.VideoCapture(r'.\Image-Processing\Video to Frames\v_SoccerPenalty_g12_c01.mp4')
success,image = vidcap.read()
count = 0
while success:
  cv2.imwrite(r".\Image-Processing\Video to Frames\Frames\frame%d.jpg" % count, image)     # save frame as JPEG file      
  success,image = vidcap.read()
  print('Read a new frame: ', success)
  count += 1
```
